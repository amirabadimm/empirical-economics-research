# Canonical monthly asset returns

This notebook audits the preserved monthly level and return panels for gold, TEDPIX, Etemad fixed income, and Tehran housing. Portfolio optimization and prior allocation results were intentionally removed pending a properly specified risk-adjusted framework.

```python
from pathlib import Path

import pandas as pd

project_root = Path.cwd()
if not (project_root / 'data').exists():
    project_root = project_root.parent

levels_path = project_root / 'data/processed/analysis/monthly_asset_levels.csv'
returns_path = project_root / 'data/processed/analysis/monthly_asset_returns.csv'
levels = pd.read_csv(levels_path)
returns = pd.read_csv(returns_path)

{'levels_shape': levels.shape, 'returns_shape': returns.shape}
```

## Coverage and missing-return audit

Missing returns remain missing. They are not replaced with zero, forward-filled, or interpolated.

```python
return_value_column = 'monthly_return' if 'monthly_return' in returns.columns else 'return'
coverage = (
    returns.groupby('asset_id', as_index=False)
    .agg(
        first_period=('jalali_period', 'min'),
        last_period=('jalali_period', 'max'),
        panel_rows=('jalali_period', 'size'),
        valid_returns=(return_value_column, 'count'),
    )
)
coverage['missing_returns'] = coverage['panel_rows'] - coverage['valid_returns']
coverage
```
